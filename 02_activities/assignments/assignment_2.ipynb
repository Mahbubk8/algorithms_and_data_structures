The problem assigned to my partner Kavin requires identifying to find all possible paths from the root of a binary tree to its leaf nodes. The function takes the root of a binary tree as input and returns a list where each  sublist represents a unique path from the root to a leaf.


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


```python
from typing import List

class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

def bt_path(root: TreeNode) -> List[List[int]]:
    if not root:
        return []
    
    output = []
    paths = [(root, [root.val])]
  
    while paths:
        node, path = paths.pop()
      
        if not node.left and not node.right:
            output.append(path)
      
        if node.right is not None:
            paths.append((node.right, path + [node.right.val]))
        if node.left is not None:
            paths.append((node.left, path + [node.left.val]))
    
    return output

# New example binary tree:
#         4
#        / \
#       2   6
#      / \ / \
#     1  3 5  7

t = TreeNode(4)
t.left = TreeNode(2)
t.right = TreeNode(6)
t.left.left = TreeNode(1)
t.left.right = TreeNode(3)
t.right.left = TreeNode(5)
t.right.right = TreeNode(7)

# Call the function to get all root-to-leaf paths
print(bt_path(t))

#Explanation:
#TreeNode class: The TreeNode class is used to define each node in the binary tree, with attributes for the node's value, left child, and right child.
#bt_path function: This function takes the root node of a binary tree and returns a list of all root-to-leaf paths. It uses a depth-first search (DFS) with an iterative approach using a stack.
#Return: [[4, 2, 1], [4, 2, 3], [4, 6, 5], [4, 6, 7]]
```
