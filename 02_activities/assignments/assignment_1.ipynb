#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List[int]) -> int:
    n = len(nums)  # The length of the list represents n (the number of elements we should have)
    expected_sum = (n + 1) * (n + 2) // 2  # Sum of numbers from 1 to n+1
    actual_sum = sum(nums)  # Sum of numbers in the list
    return expected_sum - actual_sum  # The missing number
```

```python
from typing import List

def missing_num(nums: List[int]) -> int:
    n = len(nums)  # The length of the list
    xor_all = 0
    xor_nums = 0
    
    # XOR all numbers from 1 to n+1
    for i in range(1, n + 2):
        xor_all ^= i
    
    # XOR all numbers in the given list
    for num in nums:
        xor_nums ^= num
    # The missing number is the XOR of the above two results
    return xor_all ^ xor_nums    
```


-   Explain why your solution works


```python
# Your answer here
Why it cancels out numbers: The key is that every number in the list (and the full set) appears exactly once, and XOR-ing two identical numbers results in zero. So, when you XOR the complete set of numbers (1 to n+1n+1n+1) with the numbers in the list, all the numbers that appear in both sets cancel each other out, leaving only the missing number.
```
